```
from inference import MountainNER
```

```
# Initialize the model using pre-trained weights
ner = MountainNER(model_path="weights/mountain_ner_model")

# Sample input sentences containing geographical names
test_sentences = [
    "Next winter our expedition aims to climb K2.",
    "Mount Fuji is an iconic natural landmark in Japan.",
    "We compared Mount Blanc and Matterhorn before deciding our itinerary.",
    "Hoverla and Pip Ivan are famous peaks located in Ukraine.",
    "It is very dangerous to conquer Nanga Parbat during storm season."
]
```

```
print("================ MOUNTAIN NER DEMO ================\n")

# Process each sentence and extract detected mountain entities
for sentence in test_sentences:
    entities = ner.extract_mountains(sentence)
    print(f"Input Text: '{sentence}'")
    if entities:
        for ent in entities:
            print(f"Found: {ent['mountain']} (Confidence: {ent['score']*100:.2f}%)")
    else:
        print("No mountains detected.")
    print("-" * 50)
```
